```python
def find_problems_with_images(folder_paths):
    # List to store folders with 'images' subfolder
    folders_with_images = []

    # Iterate over each folder path
    for folder_path in folder_paths:
        # Check if "problems" folder exists in the current folder path
        problems_folder_path = os.path.join(folder_path, "problems")
        # Iterate over all subfolders in the "problems" folder
        for subfolder in os.listdir(problems_folder_path):
            subfolder_path = os.path.join(problems_folder_path, subfolder)
            # Check if the subfolder is a directory and has an "images" folder
            if os.path.isdir(subfolder_path) and "images" in os.listdir(subfolder_path):
                # Add to the list
                folders_with_images.append(subfolder_path)

    return folders_with_images

crawled_problems_with_images = find_problems_with_images([
    # "/home/kaixin/Desktop/mmcode/crawl/crawled/aizu",
    # "/home/kaixin/Desktop/mmcode/crawl/crawled/atcoder",
    # "/home/kaixin/Desktop/mmcode/crawl/crawled/codeforces",
    # "/home/kaixin/Desktop/mmcode/crawl/crawled/codewars",
    # "/home/kaixin/Desktop/mmcode/crawl/crawled/leetcode",
    "/home/kaixin/Desktop/mmcode/crawl/crawled/openkattis",
    # "/home/kaixin/Desktop/mmcode/crawl/crawled/project-euler"
])
```

```python
def load_crawled_problems(folder_path):
    """
    Function to load data from a JSON file.

    Parameters:
    file_path (str): The path to the 'data.json' file.

    Returns:
    dict: The data loaded from the JSON file.
    """
    with open(os.path.join(folder_path, "data.json"), 'r') as file:
        data = json.load(file)
        return data
```

```python
load_crawled_problems(crawled_problems_with_images[0])
```

```python
crawled_dataset_dict = {path: load_crawled_problems(path)["problem"] for path in crawled_problems_with_images}
```

## Matching

```python
import multiprocessing
from tqdm.notebook import tqdm
import ngram

# Initialize ngram
G = ngram.NGram(N=5)

# Define the function to compare one crawled problem with all taco problems
def compare_with_all_tacos(crawled_item):
    crawled_path, crawled_problem = crawled_item
    max_score = -1  # Initialize with a value lower than the lowest possible score
    best_match = None

    for taco_id, taco_problem in taco_dataset_dict.items():
        score = G.compare(taco_problem, crawled_problem)
        if score > max_score:  # Update max_score and best_match if current score is higher
            max_score = score
            best_match = {
                "crawled_path": crawled_path,
                "taco_id": taco_id,
                "score": score
            }

    return best_match

# Create a pool of worker processes
pool = multiprocessing.Pool(processes=8)

# List for collecting all matches
all_matches = []

# Iterate over crawled problems and process them in parallel
futures = []
for crawled_item in crawled_dataset_dict.items():
    future = pool.apply_async(compare_with_all_tacos, args=(crawled_item,))
    futures.append(future)

for future in tqdm(futures):
    all_matches.append(future.get())

# Close the pool and wait for the work to finish
pool.close()
pool.join()
```

```python
import json

# Serializing the list to JSON
with open('match.json', 'w') as file:
    json.dump(all_matches, file)
```
