```python
import os
from PIL import Image

def process(png_file_path):
    image = Image.open(png_file_path).convert("RGBA")
    new_image = Image.new("RGBA", image.size, "WHITE") # Create a white rgba background
    new_image.paste(image, (0, 0), image)
    new_image = new_image.convert('RGB')
    del image
    return new_image

def process_png_files(folder_path):
    for root, dirs, files in os.walk(folder_path):
        for file in files:
            if file.endswith(".png"):
                file_path = os.path.join(root, file)
                processed_image = process(file_path)
                processed_image.save(file_path)  # Saves the image in place

# Example usage
folder_path = '/home/kaixin/new_mmcode_processed'  # Replace with your folder path
process_png_files(folder_path)
```
