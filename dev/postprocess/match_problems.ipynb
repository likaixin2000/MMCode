```python
def find_problems_with_images(folder_paths):
    # List to store folders with 'images' subfolder
    folders_with_images = []

    # Iterate over each folder path
    for folder_path in folder_paths:
        # Check if "problems" folder exists in the current folder path
        problems_folder_path = os.path.join(folder_path, "problems")
        # Iterate over all subfolders in the "problems" folder
        for subfolder in os.listdir(problems_folder_path):
            subfolder_path = os.path.join(problems_folder_path, subfolder)
            # Check if the subfolder is a directory and has an "images" folder
            if os.path.isdir(subfolder_path) and "images" in os.listdir(subfolder_path):
                # Add to the list
                folders_with_images.append(subfolder_path)

    return folders_with_images

crawled_problems_with_images = find_problems_with_images([
    # "/home/kaixin/Desktop/mmcode/crawl/crawled/aizu",
    # "/home/kaixin/Desktop/mmcode/crawl/crawled/atcoder",
    # "/home/kaixin/Desktop/mmcode/crawl/crawled/codeforces",
    # "/home/kaixin/Desktop/mmcode/crawl/crawled/codewars",
    # "/home/kaixin/Desktop/mmcode/crawl/crawled/leetcode",
    "/home/kaixin/Desktop/mmcode/crawl/crawled/openkattis",
    # "/home/kaixin/Desktop/mmcode/crawl/crawled/project-euler"
])
```

```python
def load_crawled_problems(folder_path):
    """
    Function to load data from a JSON file.

    Parameters:
    file_path (str): The path to the 'data.json' file.

    Returns:
    dict: The data loaded from the JSON file.
    """
    with open(os.path.join(folder_path, "data.json"), 'r') as file:
        data = json.load(file)
        return data
```

## Matching

```python
import multiprocessing
from tqdm.notebook import tqdm
import ngram

# Initialize ngram
G = ngram.NGram(N=5)

# Define the function to compare one crawled problem with all taco problems
def compare_with_all_tacos(crawled_item):
    crawled_path, crawled_problem = crawled_item
    max_score = -1  # Initialize with a value lower than the lowest possible score
    best_match = None

    for taco_id, taco_problem in taco_dataset_dict.items():
        score = G.compare(taco_problem, crawled_problem)
        if score > max_score:  # Update max_score and best_match if current score is higher
            max_score = score
            best_match = {
                "crawled_path": crawled_path,
                "taco_id": taco_id,
                "score": score
            }

    return best_match

# Create a pool of worker processes
pool = multiprocessing.Pool(processes=8)

# List for collecting all matches
all_matches = []

# Iterate over crawled problems and process them in parallel
futures = []
for crawled_item in crawled_dataset_dict.items():
    future = pool.apply_async(compare_with_all_tacos, args=(crawled_item,))
    futures.append(future)

for future in tqdm(futures):
    all_matches.append(future.get())

# Close the pool and wait for the work to finish
pool.close()
pool.join()
```

```python
import json

# Serializing the list to JSON
with open('match.json', 'w') as file:
    json.dump(all_matches, file)
```

```python
def filter_duplicates(data_list):
    path_duplicates = {}
    id_duplicates = {}

    # Check for duplicates by path and id
    for item in data_list:
        path = item['crawled_path']
        taco_id = item['taco_id']

        # Check for path duplicates
        if path in path_duplicates:
            path_duplicates[path].append(item)
        else:
            path_duplicates[path] = [item]

        # Check for id duplicates
        if taco_id in id_duplicates:
            id_duplicates[taco_id].append(item)
        else:
            id_duplicates[taco_id] = [item]

    # Filter out non-duplicates
    path_duplicates = {k: v for k, v in path_duplicates.items() if len(v) > 1}
    id_duplicates = {k: v for k, v in id_duplicates.items() if len(v) > 1}

    return path_duplicates, id_duplicates

# Filter and print the duplicates
path_dups, id_dups = filter_duplicates(filtered_matches)

print("Duplicates by Crawled Path:")
for key, values in path_dups.items():
    print(f"Path: {key}")
    for value in values:
        print(value)

print("\nDuplicates by Taco ID:")
for key, values in id_dups.items():
    print(f"ID: {key}")
    for value in values:
        print(value)
```

```python
def remove_duplicates(data_list):
    best_by_path = {}
    best_by_id = {}

    # First pass for path
    for item in data_list:
        path = item['crawled_path']
        if path not in best_by_path or (item['score'] > best_by_path[path]['score'] or
                                        (item['score'] == best_by_path[path]['score'] and
                                         item['crawled_path'] > best_by_path[path]['crawled_path'])):
            best_by_path[path] = item

    # Second pass for id
    for item in data_list:
        taco_id = item['taco_id']
        if taco_id not in best_by_id or (item['score'] > best_by_id[taco_id]['score'] or
                                         (item['score'] == best_by_id[taco_id]['score'] and
                                          item['crawled_path'] > best_by_id[taco_id]['crawled_path'])):
            best_by_id[taco_id] = item

    # Combine and get unique items
    unique_items = {item['taco_id']: item for item in best_by_path.values()}
    unique_items.update({item['taco_id']: item for item in best_by_id.values()})

    return list(unique_items.values())

# Remove duplicates and get the unique items
deduped_filtered_data = remove_duplicates(filtered_matches)

len(deduped_filtered_data)
```

## Postprocess

```python
def get_taco_data(taco_id):
    if taco_id.startswith("train_"):
        real_id = int(taco_id[6:])
        return taco_dataset["train"][real_id]
    elif taco_id.startswith("test_"):
        real_id = int(taco_id[5:])
        return taco_dataset["test"][real_id]
    else:
        raise ValueError()
```

```python
# Copy files to new dataset
import shutil
import os
import json

destination_folder="/home/kaixin/Desktop/mmcode/mmcode_dataset"
for item in deduped_filtered_data:
    taco_id = item["taco_id"]
    crawled_path = item["crawled_path"]

    if "codeforces" in crawled_path or "aizu" in crawled_path:
        continue

    # Use the last folder name of the crawled_path as the new folder name
    new_folder_name = os.path.basename(os.path.normpath(crawled_path))
    new_folder_path = os.path.join(destination_folder, new_folder_name)

    # Create the new folder
    os.makedirs(new_folder_path, exist_ok=True)

    # Copy files from crawled_path to the new folder
    shutil.copytree(crawled_path, new_folder_path, dirs_exist_ok=True)

    # write taco data to a file
    taco_data = get_taco_data(taco_id)
    taco_json_path = os.path.join(new_folder_path, "taco.json")
    with open(taco_json_path, 'w') as f:
        json.dump(taco_data, f, indent=4)

    
```
