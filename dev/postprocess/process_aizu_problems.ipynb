```python
import os
def get_problems_with_images(problems_folder):
    """
    Scan the image folder to retrieve problems with images.
    """

    # Get all subdirectories with image folder
    subdirs = [os.path.join(problems_folder, d) for d in os.listdir(problems_folder) 
               if os.path.isdir(os.path.join(problems_folder, d)) and os.path.exists(os.path.join(problems_folder, d, "images"))]

    # Return the list of strs
    return subdirs
```

```python
probs_w_image = get_problems_with_images("/home/kaixin/Desktop/mmcode/crawl/crawled/aizu/problems")
len(probs_w_image)
```

## Copy files

```python
# Copy files to new dataset
import shutil
import os
import json

cnt = 0
destination_folder="/home/kaixin/Desktop/mmcode/mmcode_dataset"
for crawled_path in probs_w_image:
    with open(os.path.join(crawled_path, 'data.json'), 'r') as f:
        problem_data = json.load(f)

    solution_path = os.path.join(crawled_path, "submissions", "python", "OK.json")
    if os.path.exists(solution_path):
        with open(solution_path, 'r') as f:
            solutions = json.loads(f.read())
            solutions = json.dumps([sol["code"] for sol in solutions])
    else:
        solutions = json.dumps([])

    io_path = os.path.join(crawled_path, "input_output.json")
    if os.path.exists(io_path):
        with open(io_path, 'r') as f:
            _ios = json.loads(f.read())
        if len(_ios) == 0:
            continue
        ios = {"inputs": [], "outputs": []}
        for s in _ios:
            ios["inputs"].append(s["input"])
            ios["outputs"].append(s["output"])
        ios = json.dumps(ios)
    else:
        continue

    problem_data["input_output"] = ios
    problem_data["solutions"] = solutions

    problem_data["question"] = problem_data.pop("problem")
    problem_data["starter_code"] = ""
    problem_data["difficulty"] = None
    problem_data["source"] = "aizu"
    problem_data["name"] = problem_data.pop("problem_title")

    # Use the last folder name of the crawled_path as the new folder name
    new_folder_name = os.path.basename(os.path.normpath(crawled_path))
    new_folder_path = os.path.join(destination_folder, new_folder_name)

    # Create the new folder
    os.makedirs(new_folder_path, exist_ok=True)

    # Copy files from crawled_path to the new folder
    shutil.copytree(os.path.join(crawled_path, "images"), os.path.join(new_folder_path, "images"), dirs_exist_ok=True)

    with open(os.path.join(new_folder_path, "data.json"), 'w') as f:
        json.dump(problem_data, f, indent=4)
    cnt += 1

print(cnt)
```
