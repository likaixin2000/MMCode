```python
import os
import shutil

def delete_empty_image_subfolders(directory):
    for root, dirs, files in os.walk(directory, topdown=False):
        for dir_name in dirs:
            dir_path = os.path.join(root, dir_name)
            if dir_name.lower() == "images" and len(os.listdir(dir_path)) == 0:
                parent_dir = os.path.dirname(dir_path)
                shutil.rmtree(parent_dir)
                print(f"Deleted empty subfolder: {parent_dir}")

# Replace 'your_directory_path' with the path to the folder you want to search
directory_path = r'C:\Users\user\Desktop\mmcode_dataset'
delete_empty_image_subfolders(directory_path)
```

## Reorder images

```python
import os
import json
import re

def update_json_and_rearrange_images(directory):
    for root, dirs, files in os.walk(directory):
        if 'images' not in dirs:
            continue

        images_dir = os.path.join(root, 'images')
        images = [int(os.path.splitext(img)[0]) for img in os.listdir(images_dir) if img.lower().endswith('.png')]

        # Update data.json 
        json_path = os.path.join(root, 'data.json')
        with open(json_path, 'r+') as json_file:
            data = json.load(json_file)
            question = data['question']
            updated_tokens = False

            # Find markdown image tokens and update if image exists
            img_counter = 1
            new_question = question
            for i, match in enumerate(re.finditer(r'!\[image\]\((.*?)\)', question), start=1):

                if i in images:
                    new_name = f"{img_counter}.png"
                    new_question = new_question.replace(match.group(0), rf'![image]({img_counter}.png)')
                    if i != img_counter:
                        # Rename the file
                        os.rename(os.path.join(images_dir, f"{i}.png"), os.path.join(images_dir, new_name))
                    updated_tokens = True
                    img_counter += 1
                else:
                    new_question = new_question.replace(match.group(0), '')

            if updated_tokens:
                data['question'] = new_question
                json_file.seek(0)
                json_file.truncate()
                json.dump(data, json_file)
                print(f"Updated data.json in {root}")


# Replace 'your_directory_path' with the path to the folder you want to search
directory_path = r"C:\Users\user\Desktop\mmcode_dataset"
update_json_and_rearrange_images(directory_path)
```
