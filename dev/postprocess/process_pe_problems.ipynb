```python
import os
def get_problems_with_images(problems_folder):
    """
    Scan the image folder to retrieve problems with images.
    """

    # Get all subdirectories with image folder
    subdirs = [os.path.join(problems_folder, d) for d in os.listdir(problems_folder) 
               if os.path.isdir(os.path.join(problems_folder, d)) and os.path.exists(os.path.join(problems_folder, d, "images"))]

    # Return the list of strs
    return subdirs
```

```python
def convert_txt_to_dict(file_path):
    """
    Convert the contents of a text file with a specific format into a dictionary.
    The file format is expected to have lines with a number, a dot, and a value.
    Example line format: '1. 233168'
    
    Args:
    file_path (str): The path to the text file.
    
    Returns:
    dict: A dictionary with the number as the key and the value as the dictionary value.
    """
    data_dict = {}
    with open(file_path, 'r') as file:
        for line in file:
            # Splitting each line at the dot and stripping whitespace
            parts = line.strip().split('.')
            if len(parts) == 2:
                key, value = parts[0].strip(), parts[1].strip()
                # Converting key and value to appropriate types
                data_dict[int(key)] = value
    return data_dict

answers = convert_txt_to_dict("/home/kaixin/Desktop/mmcode/crawl/crawled/project-euler/answers.txt")

```

```python
import glob
import re

def remove_first_comment_block(directory):
    for filepath in glob.glob(f"{directory}/p*.py"):
        filename = os.path.basename(filepath)
        if filename.startswith('p') and filename[1:-3].isdigit():
            with open(filepath, 'r') as file:
                lines = file.readlines()

            in_comment_block = True
            new_lines = []

            for line in lines:
                if in_comment_block:
                    if line.startswith('#') or line.strip() == '':
                        continue
                    else:
                        in_comment_block = False
                new_lines.append(line)
            print(new_lines)
            with open(filepath, 'w') as file:
                file.writelines(new_lines)

# Example usage
remove_first_comment_block('/home/kaixin/Desktop/mmcode/crawl/crawled/project-euler/python')
```

```python
import glob

def convert_main_block(directory):
    change_count = 0  # Initialize a counter for changes

    for filepath in glob.glob(f"{directory}/*.py"):
        with open(filepath, 'r') as file:
            content = file.read()

        # Define the patterns
        old_pattern = "if __name__ == \"__main__\":\n\tprint(compute())"
        new_pattern = "print(compute())"

        # Replace the pattern and count the change
        if old_pattern in content:
            modified_content = content.replace(old_pattern, new_pattern)
            with open(filepath, 'w') as file:
                file.write(modified_content)
            change_count += 1  # Increment the change counter

    return change_count

# Example usage
num_changes = convert_main_block('/home/kaixin/Desktop/mmcode/crawl/crawled/project-euler/python')
print(f"Number of files changed: {num_changes}")
```

```python
# Copy files to new dataset
import shutil
import os
import json

cnt = 0
destination_folder="/home/kaixin/Desktop/mmcode/mmcode_dataset"
for crawled_path in pe_probs_w_image:
    with open(os.path.join(crawled_path, 'data.json'), 'r') as f:
        problem_data = json.load(f)
    problem_id = int(crawled_path.split('/ep_')[-1])

    solution_path = os.path.join("/home/kaixin/Desktop/mmcode/crawl/crawled/project-euler/python", f"p{problem_id:03d}.py")
    if os.path.exists(solution_path):
        with open(solution_path, 'r') as f:
            solution = json.dumps(f.read())
    else:
        solution = json.dumps([])

    output = answers.get(int(problem_id), None)
    if not output:
        continue
    problem_data["input_output"] = json.dumps({"inputs": "", "outputs": output})
    problem_data["solutions"] = solution

    # Use the last folder name of the crawled_path as the new folder name
    new_folder_name = os.path.basename(os.path.normpath(crawled_path))
    new_folder_path = os.path.join(destination_folder, new_folder_name)

    # Create the new folder
    os.makedirs(new_folder_path, exist_ok=True)

    # Copy files from crawled_path to the new folder
    shutil.copytree(crawled_path, new_folder_path, dirs_exist_ok=True)

    with open(os.path.join(new_folder_path, "data.json"), 'w') as f:
        json.dump(problem_data, f, indent=4)
    cnt += 1

print(cnt)

    
```
